```python
states = ["S0", "S1", "S3", "S4", "G"]

actions = ["Up", "Down", "Left", "Right"]

# Discount factor
GAMMA = 0.9

# Rewards
NORMAL_REWARD = -1
GOAL_REWARD = 10
OBSTACLE_REWARD = -5

# 2. POLICY GIVEN IN THE QUESTION

policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}

# 3. TRANSITION FUNCTION

def transition(state, action):

    # S0 = (0,0)
    if state == "S0":

        if action == "Right":
            return "S1"

        elif action == "Down":
            return "S0"      # X is obstacle

        elif action == "Up":
            return "S0"      # Outside grid

        elif action == "Left":
            return "S0"      # Outside grid
    # S1 = (0,1)
    elif state == "S1":

        if action == "Right":
            return "G"

        elif action == "Down":
            return "S3"

        elif action == "Up":
            return "S1"

        elif action == "Left":
            return "S0"
    # S3 = (1,1)
    elif state == "S3":

        if action == "Right":
            return "S4"

        elif action == "Up":
            return "S1"

        elif action == "Down":
            return "S3"      # Outside grid

        elif action == "Left":
            return "S3"      # X is obstacle
    # S4 = (1,2)
    elif state == "S4":

        if action == "Up":
            return "S1"

        elif action == "Left":
            return "S3"

        elif action == "Right":
            return "S4"      # Outside grid

        elif action == "Down":
            return "S4"      # Outside grid


    # Goal state
    elif state == "G":
        return "G"
# 4. REWARD FUNCTION

def get_reward(next_state):

    if next_state == "G":
        return GOAL_REWARD

    else:
        return NORMAL_REWARD

# 5. STATE VALUE FUNCTION

def state_value_function(state):

    current_state = state

    total_return = 0

    discount = 1


    # Calculate discounted return
    for step in range(1000):

        # Stop if goal is reached
        if current_state == "G":
            break

        # Get action from policy
        action = policy[current_state]

        # Find next state
        next_state = transition(current_state, action)

        # Get reward
        reward = get_reward(next_state)

        # Add discounted reward
        total_return = total_return + discount * reward

        # Update discount
        discount = discount * GAMMA

        # Move to next state
        current_state = next_state

    return total_return
# 6. ACTION VALUE FUNCTION

def action_value_function(state, action):

    # Take the action
    next_state = transition(state, action)

    # Get immediate reward
    reward = get_reward(next_state)

    # If goal is reached, there is no future reward
    if next_state == "G":
        return reward

    # Calculate future value
    future_value = state_value_function(next_state)

    # Q(s,a) = R + gamma * V(next state)
    q_value = reward + GAMMA * future_value

    return q_value

# 7. DISPLAY POLICY

print("=" * 60)
print("POLICY")
print("=" * 60)

for state, action in policy.items():
    print(state, "→", action)
# 8. CALCULATE STATE VALUES

print("\n" + "=" * 60)
print("STATE VALUES")
print("=" * 60)

for state in ["S0", "S1", "S3", "S4"]:

    value = state_value_function(state)

    print("V(" + state + ") =", round(value, 4))

# 9. CALCULATE ACTION VALUE

print("\n" + "=" * 60)
print("ACTION VALUE")
print("=" * 60)

q_value = action_value_function("S0", "Up")

print("Q(S0, Up) =", round(q_value, 4))
# 10. FINAL ANSWERS

print("\n" + "=" * 60)
print("FINAL ANSWERS")
print("=" * 60)

print("State Value of S0 =", round(state_value_function("S0"), 4))

print("Action Value of S0 for Up =",
      round(action_value_function("S0", "Up"), 4))
```
